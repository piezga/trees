```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

# Try to import Tesseract, but handle the error if it's missing
try:
    import pytesseract
    from pytesseract import TesseractNotFoundError
    TESSERACT_AVAILABLE = True
except ImportError:
    TESSERACT_AVAILABLE = False
    print("Warning: Pytesseract library not found (OCR disabled).")

def analyze_image_layout(image, inset=8):
    """ 
    Separates the Main Map from the Colorbar Legend.
    
    Args:
    - inset: Number of pixels to shave off the borders of the detected map 
             to remove the black axis lines.
    """
    gray = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)
    _, thresh = cv2.threshold(gray, 220, 255, cv2.THRESH_BINARY_INV)
    
    contours, _ = cv2.findContours(thresh, cv2.RETR_EXTERNAL, cv2.CHAIN_APPROX_SIMPLE)
    
    map_roi = None
    legend_roi = None
    total_area = image.shape[0] * image.shape[1]

    for cnt in contours:
        x, y, w, h = cv2.boundingRect(cnt)
        area = w * h
        aspect_ratio = w / float(h)
        
        # Heuristic 1: The MAP is the largest box, usually wide (BCI is 2:1 ratio)
        if area > 0.3 * total_area and aspect_ratio > 1.2:
            # Apply INSET to remove the black axis lines
            # Check bounds to avoid negative indexing
            y_start = min(y + inset, y + h)
            y_end = max(y + h - inset, y)
            x_start = min(x + inset, x + w)
            x_end = max(x + w - inset, x)
            
            map_roi = image[y_start:y_end, x_start:x_end]
            print(f"Found Map: {w}x{h} (Cropped with {inset}px inset)")
            
        # Heuristic 2: The COLORBAR is tall and thin
        elif area > 0.01 * total_area and aspect_ratio < 0.3:
            legend_roi = image[y:y+h, x:x+w]
            print(f"Found Legend: {w}x{h}")

    return map_roi, legend_roi

def get_min_max_from_legend(legend_img):
    """ Attempts to read text. Returns None if Tesseract is missing/fails. """
    if not TESSERACT_AVAILABLE or legend_img is None:
        return None, None

    try:
        gray = cv2.cvtColor(legend_img, cv2.COLOR_BGR2GRAY)
        scale = 3 
        gray = cv2.resize(gray, None, fx=scale, fy=scale, interpolation=cv2.INTER_CUBIC)
        _, binary = cv2.threshold(gray, 150, 255, cv2.THRESH_BINARY + cv2.THRESH_OTSU)

        # Try to run OCR
        data = pytesseract.image_to_data(binary, config='--psm 6', output_type=pytesseract.Output.DICT)
        
        numbers = []
        for i, text in enumerate(data['text']):
            clean_text = ''.join(c for c in text if c.isdigit() or c == '.')
            if clean_text:
                try:
                    val = float(clean_text)
                    numbers.append((data['top'][i], val))
                except ValueError:
                    continue
        
        if not numbers: return None, None
        
        numbers.sort(key=lambda x: x[0])
        return numbers[-1][1], numbers[0][1] # Bottom (Min), Top (Max)

    except Exception as e:
        print(f">> OCR failed: {e}")
        return None, None

def extract_bci_matrix(image_path, target_val, interval, 
                       manual_min=None, manual_max=None,
                       grid_rows=25, grid_cols=50):
    
    img = cv2.imread(image_path)
    if img is None: 
        print("Error: Image not found.")
        return None, None

    # 1. ANALYZE LAYOUT (With Safety Inset)
    # inset=8 removes roughly 8 pixels from each side of the black box
    map_img, legend_img = analyze_image_layout(img, inset=5)
    
    if map_img is None:
        print("Could not auto-detect map layout. Using full image.")
        map_img = img
        legend_img = img 

    # 2. DETERMINE VALUES (OCR vs MANUAL)
    min_val, max_val = get_min_max_from_legend(legend_img)
    
    # Override if OCR failed or User provided manuals
    if min_val is None or (manual_min is not None):
        if manual_min is None or manual_max is None:
            print("!!! CRITICAL: OCR failed and no manual values provided.")
            return None, None
        min_val = manual_min
        max_val = manual_max
        print(f"Using Manual Calibration: Min={min_val}, Max={max_val}")
    else:
        print(f"OCR Auto-Calibration: Min={min_val}, Max={max_val}")

    # 3. DETECT INVERSION (Dark vs Light)
    # Heuristic: Check intensity at top vs bottom of legend
    legend_gray = cv2.cvtColor(legend_img, cv2.COLOR_BGR2GRAY)
    h, w = legend_gray.shape
    top_bright = np.mean(legend_gray[0:int(h*0.1), :])
    bot_bright = np.mean(legend_gray[int(h*0.9):, :])
    
    # If top (Max val) is darker than bottom (Min val) -> Invert
    invert = top_bright < bot_bright
    if invert: print("Mode: Darker colors = Higher Values")
    else: print("Mode: Brighter colors = Higher Values")
    
    # 4. CREATE MATRIX
    gray_map = cv2.cvtColor(map_img, cv2.COLOR_BGR2GRAY)
    resized = cv2.resize(gray_map, (grid_cols, grid_rows), interpolation=cv2.INTER_AREA)
    norm_grid = resized / 255.0
    
    if invert:
        # Dark = Max
        physical_grid = max_val - (norm_grid * (max_val - min_val))
    else:
        # Bright = Max
        physical_grid = min_val + (norm_grid * (max_val - min_val))
        
    lower = target_val - interval
    upper = target_val + interval
    binary_matrix = ((physical_grid >= lower) & (physical_grid <= upper)).astype(int)
    
    return binary_matrix, physical_grid

# --- HOW TO RUN ---

# Replace with your values found on the map's colorbar
MANUAL_MIN = 3.6  # Example: Bottom number on legend
MANUAL_MAX = 5.4  # Example: Top number on legend

matrix, continuous_map = extract_bci_matrix(
    image_path='bci-ph.png', 
    target_val=5.2, 
    interval=0.1,
    manual_min=MANUAL_MIN, 
    manual_max=MANUAL_MAX
)

if matrix is not None:
    plt.figure(figsize=(10, 5))
    
    # Check that borders are clean (no black frame)
    plt.subplot(1, 2, 1)
    plt.imshow(continuous_map, cmap='viridis')
    plt.title("Reconstructed Map (Check Edges)")
    plt.colorbar()
    
    plt.subplot(1, 2, 2)
    plt.imshow(matrix, cmap='binary')
    plt.title("Selected Area")
    plt.show()
    
    print("Matrix Shape:", matrix.shape)
```

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
from scipy.spatial import distance

def analyze_image_layout(image, inset=12):
    """ Separates Map from Legend (with safety inset). """
    gray = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)
    _, thresh = cv2.threshold(gray, 230, 255, cv2.THRESH_BINARY_INV)
    contours, _ = cv2.findContours(thresh, cv2.RETR_EXTERNAL, cv2.CHAIN_APPROX_SIMPLE)
    
    map_roi = None
    legend_roi = None
    total_area = image.shape[0] * image.shape[1]

    for cnt in contours:
        x, y, w, h = cv2.boundingRect(cnt)
        area = w * h
        aspect_ratio = w / float(h)
        
        # MAP: Large and Wide
        if area > 0.3 * total_area and aspect_ratio > 1.2:
            y_start, y_end = min(y + inset, y + h), max(y + h - inset, y)
            x_start, x_end = min(x + inset, x + w), max(x + w - inset, x)
            map_roi = image[y_start:y_end, x_start:x_end]
            
        # LEGEND: Tall and Thin
        elif area > 0.01 * total_area and aspect_ratio < 0.3:
            legend_roi = image[y:y+h, x:x+w]

    return map_roi, legend_roi

def build_color_scale(legend_img, min_val, max_val):
    """ Scans the center of the legend to build the color dictionary. """
    h, w, _ = legend_img.shape
    center_x = w // 2
    # Scan vertically
    legend_colors = legend_img[:, center_x, :] 
    # Linspace values from Top (Max) to Bottom (Min)
    legend_values = np.linspace(max_val, min_val, h)
    return legend_colors, legend_values

def map_colors_to_values(map_img, legend_colors, legend_values, rows=25, cols=50):
    """
    Matches pixels to legend.
    CRITICAL FIX: Filters out background pixels that are 'too far' from any legend color.
    """
    # 1. Resize
    grid_img = cv2.resize(map_img, (cols, rows), interpolation=cv2.INTER_NEAREST)
    map_pixels = grid_img.reshape(-1, 3)
    
    # 2. Calculate Distances to ALL legend colors
    dists = distance.cdist(map_pixels, legend_colors, metric='euclidean')
    
    # 3. Find Closest Matches
    nearest_indices = np.argmin(dists, axis=1)
    min_distances = np.min(dists, axis=1)
    
    # 4. Filter Background (The "Yellow Line" Fix)
    # If the closest matching color is still 50 units away (e.g. White vs Yellow), it's background.
    # We set these to NaN so they don't register as "Max Value"
    valid_mask = min_distances < 60.0 # Threshold (adjust if needed)
    
    flat_physical = legend_values[nearest_indices]
    
    # Apply the mask (Set invalid pixels to a dummy value like -999)
    flat_physical[~valid_mask] = -999
    
    physical_grid = flat_physical.reshape(rows, cols)
    
    return physical_grid

def process_nitrogen_map(image_path, target_val, interval, manual_min, manual_max):
    img = cv2.imread(image_path)
    if img is None: return None, None
    
    # 1. Split Layout
    map_roi, legend_roi = analyze_image_layout(img, inset=12)
    if map_roi is None: return None, None

    # 2. Learn Colors
    legend_colors, legend_values = build_color_scale(legend_roi, manual_min, manual_max)
    
    # 3. Map Pixels
    physical_grid = map_colors_to_values(map_roi, legend_colors, legend_values)
    
    # 4. Create Binary Selection
    lower = target_val - interval
    upper = target_val + interval
    
    # Select only Valid pixels (-999 will fail this check automatically)
    binary_matrix = ((physical_grid >= lower) & (physical_grid <= upper)).astype(int)
    
    # Prepare Plotting Version (Convert -999 to NaN for transparency)
    plot_grid = physical_grid.copy()
    plot_grid[plot_grid == -999] = np.nan
    
    return binary_matrix, plot_grid

# --- RUN ---
# Update these with your exact legend numbers
MIN_VAL = 3.6 
MAX_VAL = 5.4

matrix, continuous_map = process_nitrogen_map(
    'bci-ph.png', 
    target_val=5.2, 
    interval=.1, 
    manual_min=MIN_VAL, 
    manual_max=MAX_VAL
)

if matrix is not None:
    plt.figure(figsize=(12, 5))
    
    plt.subplot(1, 2, 1)
    # Set background color to White to make NaNs obvious
    cmap = plt.cm.get_cmap('viridis').copy()
    cmap.set_bad('white') 
    plt.imshow(continuous_map, cmap=cmap)
    plt.title("pH Map (Background Removed)")
    plt.colorbar()
    
    plt.subplot(1, 2, 2)
    plt.imshow(matrix, cmap='binary')
    plt.title("Binary Selection")
    plt.show()
```

```python
import numpy as np
import pandas as pd

def calculate_overlap(binary_matrix, tree_data):
    """
    Calculates the percentage of trees that fall within the selected soil regions.
    
    Args:
    - binary_matrix: 25x50 numpy array (1=Selected Soil, 0=Background)
    - tree_data: DataFrame with columns 'gx' and 'gy' (meters).
                 BCI Grid: X (0-1000m), Y (0-500m)
                 
    Returns:
    - percentage: % of trees in the selection.
    - inside_trees: DataFrame subset of trees that are inside.
    """
    
    # 1. Extract Coordinates
    x = tree_data['gx'].values
    y = tree_data['gy'].values
    
    # 2. Convert Meters to Matrix Indices (20m grid)
    # Note: We use np.clip to ensure trees exactly on the edge (e.g., 1000m) don't crash the code.
    col_idx = np.clip((x / 20).astype(int), 0, 49)
    
    # IMPORTANT: Invert Y-axis. 
    # Matrix Row 0 is the "Top" (500m North), Row 24 is "Bottom" (0m South)
    row_idx = np.clip(((500 - y) / 20).astype(int), 0, 24)
    
    # 3. Check the Matrix
    # We look up the value (0 or 1) for every tree's location
    # This creates an array of 1s (Inside) and 0s (Outside)
    status = binary_matrix[row_idx, col_idx]
    
    # 4. Calculate Percentage
    total_trees = len(tree_data)
    count_inside = np.sum(status == 1)
    
    percentage = (count_inside / total_trees) * 100
    
    # 5. Filter Data (Optional: to see WHICH trees are inside)
    inside_trees = tree_data[status == 1]
    
    return percentage, inside_trees

# --- EXAMPLE USAGE ---

# 1. Load your tree data (Replace with your actual file)
# df_trees = pd.read_csv("my_bci_trees.csv")

# For demonstration, I will generate random trees
# Some in the "Selected" area, some outside
N = 1000
df_trees = pd.DataFrame({
    'species': ['Tree_A'] * N,
    'gx': np.random.uniform(0, 1000, N), # Random X (0-1000m)
    'gy': np.random.uniform(0, 500, N)   # Random Y (0-500m)
})

# 2. Run Calculation
# (Assuming 'matrix' is the binary matrix from your Soil Analysis code)
overlap_pct, valid_trees = calculate_overlap(matrix, df_trees)

print(f"Total Trees: {len(df_trees)}")
print(f"Trees in Selected Soil: {len(valid_trees)}")
print(f"Overlap: {overlap_pct:.2f}%")

# 3. Visual Verification
import matplotlib.pyplot as plt

plt.figure(figsize=(12, 6))

# Plot the Soil Matrix as background
plt.imshow(matrix, cmap='Greys', extent=[0, 1000, 0, 500], alpha=0.3, origin='upper')

# Plot ALL trees (Grey dots)
plt.scatter(df_trees['gx'], df_trees['gy'], c='grey', s=5, alpha=0.5, label='All Trees')

# Plot INSIDE trees (Green dots)
plt.scatter(valid_trees['gx'], valid_trees['gy'], c='green', s=5, label='Selected Trees')

plt.title(f"Overlap Analysis: {overlap_pct:.1f}% of trees are in the zone")
plt.xlabel("East-West (meters)")
plt.ylabel("North-South (meters)")
plt.legend()
plt.show()
```
