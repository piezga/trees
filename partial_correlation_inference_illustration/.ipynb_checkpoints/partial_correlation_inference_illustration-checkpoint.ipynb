## illustration of the inference of the single-subject (noise-cleaned) covariance ($C$), precision ($J$) and partial correlation ($\Pi$) matrices

```python
# X is the data (number of samples T (i.e. times) times number-of-features (N) )
def regolarizza(X,method,traintestfraction = 0.95,M0=None,standardize=False,verbose=False):

    T,N = np.shape(X)
    Ttr = int(traintestfraction*T)
    Xtr = np.copy(X[:Ttr])
    Xte = np.copy(X[Ttr:])
    

    if standardize:
        my.standardize_series(Xtr,Xte)

    if method == 'GS':
        if M0 is None:
            M0=np.eye(N)
        res = estimators.fit_Shrinkage_biasedmatrix_CV(Xtr,Xte, M0 = M0, cv_scoring="likelihood", n_jobs=None)
        best_param = res['best_params']['alpha']

    if method == 'GS_noCV':
        if M0 is None:
            M0=np.eye(N)
            
        res = estimators.fit_Shrinkage_biasedmatrix(Xtr, Xte, 0.07, M0)
        best_param = 0.07

    if method == 'RIE':
        res = estimators.fit_RIE_CV(Xtr,Xte)
        best_param = res['best_params']['eta']
        #res = estimators.fit_RIE(sub, sub[-1:])
    if method == 'cautPCA':
        res = estimators.fit_ConservativePCA_CV(Xtr,Xte)
        best_param = res['best_params']['p']        
    if method == 'LS':
        res = estimators.fit_Shrinkage_biasedmatrix_CV(Xtr,Xte, M0 = np.eye(N), cv_scoring="likelihood", n_jobs=None)
        best_param = res['best_params']['alpha']
    
    if method == 'raw':
        T,N = np.shape(sub)
        thisC=Xtr.T @ Xtr/T

    return res['test_likelihood'],res['Cclean'],best_param
```
